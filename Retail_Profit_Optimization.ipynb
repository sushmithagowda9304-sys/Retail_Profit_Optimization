```python
discount_profit = (
    df.groupby("Discount")["Profit"]
      .mean()
      .sort_index()
)

plt.figure(figsize=(10,6))

discount_profit.plot(marker="o")

plt.title("Average Profit by Discount Level")
plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.grid(True)

plt.show()
```

```python
df[["Discount","Profit"]].corr()
```

Business Question

How does discounting affect profitability, and what pricing strategy should the company adopt to maximize profits?

Answer

The analysis shows a moderate negative relationship between discounts and profit (correlation = -0.316). As discount levels increase, average profit generally declines, and very high discounts are frequently associated with financial losses.

Key Findings

There is a moderate negative correlation (-0.316) between Discount and Profit.
Orders with little or no discount generally generate higher profits.

Average profit declines as discount levels increase.
High discounts (around 50% and above) are frequently associated with negative profits.
Very high discounts (80–85%) produce the largest average losses.

Business Insights

The current discount strategy appears to erode profitability. While discounts may stimulate demand, excessive discounting significantly reduces profit margins and, in many cases, results in loss-making transactions. This indicates that the company should optimize discounts based on profitability rather than applying large, broad-based promotions.

Recommendations

Replace blanket discounts with targeted, customer- and product-specific promotions.
Limit discounts on high-margin and premium products.
Require profitability analysis before approving large discount campaigns.
Use customer segmentation to offer discounts only where they are likely to increase long-term customer value.
Continuously monitor the impact of discounts on both sales and profit through executive dashboards.

Expected Business Impact

Adopting a data-driven pricing strategy can improve profit margins, reduce loss-making transactions, optimize promotional spending, and increase overall financial performance without relying on excessive discounting.

```python
sales_segment = (
    df.groupby("Segment")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

sales_segment.plot(kind="bar", color="royalblue")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
profit_segment = (
    df.groupby("Segment")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

profit_segment.plot(kind="bar", color="green")

plt.title("Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

Business Question

Which customer segment contributes the highest sales and profit, and how should the company prioritize its customer strategy?

Answer

The Consumer segment generates the highest sales and profit, making it the company's most valuable customer segment. The Corporate segment ranks second, while the Home Office segment contributes the least to both sales and profitability.

Key Findings

Consumer segment contributes the highest sales.
Consumer segment also generates the highest profit.
Corporate is the second-largest contributor to both revenue and profit.
Home Office is the smallest contributor across both metrics.
Sales and profit rankings are consistent across all customer segments.

Business Insights

The Consumer segment is the primary driver of business growth and profitability, indicating strong demand and purchasing frequency. Since the highest sales also translate into the highest profits, investing in customer acquisition and retention within this segment is likely to deliver the greatest return. The Corporate segment represents an opportunity for further expansion through business partnerships, while the Home Office segment may require targeted marketing or specialized product offerings to increase engagement.

Recommendations

Increase marketing investment and promotional campaigns targeting Consumer customers.
Expand loyalty and retention programs for Consumer customers to maximize customer lifetime value.
Strengthen B2B sales initiatives to grow the Corporate segment.
Develop personalized offers and bundled solutions for Home Office customers to increase purchases.
Use customer segmentation to tailor pricing, promotions, and product recommendations.

Business Impact

Implementing these strategies can increase customer retention, improve customer lifetime value, optimize marketing spend, strengthen Corporate sales, and drive sustainable revenue and profit growth across customer segment

```python
import pandas as pd
import matplotlib.pyplot as plt

segment_margin = df.groupby("Segment").agg({
    "Sales":"sum",
    "Profit":"sum"
})

segment_margin["Profit Margin (%)"] = (segment_margin["Profit"] / segment_margin["Sales"]) * 100

print(segment_margin)

plt.figure(figsize=(6,5))
plt.bar(segment_margin.index, segment_margin["Profit Margin (%)"], color="green")
plt.title("Profit Margin by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Profit Margin (%)")
plt.show()
```
